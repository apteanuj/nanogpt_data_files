```python
import numpy as np
from numpy.linalg import norm, svd

rng = np.random.default_rng(0)

def rand_orthonormal_cols(m, n, rng):
    A = rng.standard_normal((m, n))
    Q, _ = np.linalg.qr(A)
    return Q[:, :n]

def power_spectral_norm(A, iters=6):
    m, n = A.shape
    if m >= n:
        v = rng.standard_normal(n); v /= (norm(v)+1e-12)
        for _ in range(iters):
            u = A @ v; u /= (norm(u)+1e-12)
            v = A.T @ u; v /= (norm(v)+1e-12)
        return abs(float(u @ (A @ v))) + 1e-12
    else:
        u = rng.standard_normal(m); u /= (norm(u)+1e-12)
        for _ in range(iters):
            v = A.T @ u; v /= (norm(v)+1e-12)
            u = A @ v; u /= (norm(u)+1e-12)
        return abs(float(u @ (A @ v))) + 1e-12

def ns_polar_tall(A, tol=1e-6, max_steps=80):
    m, n = A.shape
    I = np.eye(n, dtype=A.dtype)
    s = power_spectral_norm(A, iters=6)
    X = A / s
    for k in range(1, max_steps+1):
        XtX = X.T @ X
        err = norm(I - XtX, 'fro') / n
        if err < tol: return X, k, err
        X = 0.5 * X @ (3*I - XtX)
    return X, max_steps, err

def ns_polar_square(H, tol=1e-6, max_steps=80):
    n = H.shape[0]
    I = np.eye(n, dtype=H.dtype)
    s = power_spectral_norm(H, iters=6)
    X = H / s
    for k in range(1, max_steps+1):
        XtX = X.T @ X
        err = norm(I - XtX, 'fro') / n
        if err < tol: return X, k, err
        X = 0.5 * X @ (3*I - XtX)
    return X, max_steps, err

def polar_ground_truth(A):  # for alignment only
    U, s, Vt = svd(A, full_matrices=False)
    return U @ Vt

def run_trial(m=512, n=128, cond_min=1e-6, noise_level=0.02, tol=1e-6):
    # Old seed with wide spectrum
    U = rand_orthonormal_cols(m,n,rng)
    V = rand_orthonormal_cols(n,n,rng)
    sig = np.logspace(np.log10(cond_min), 0.0, n)
    S_old = U @ (sig[:,None] * V.T)

    # Old polar (one-time NS; in practice you cached this from a refresh)
    Q_old, _, _ = ns_polar_tall(S_old, tol=min(1e-8, tol*0.1), max_steps=120)

    # New seed with scaled Gaussian noise
    G = rng.standard_normal((m,n))
    G *= (noise_level * norm(S_old,'fro') / (norm(G,'fro') + 1e-12))
    S_new = S_old + G

    # Baseline: NS on big S_new
    Q_full, steps_big, err_big = ns_polar_tall(S_new, tol=tol, max_steps=120)

    # Relative map: H = Q_old^T S_new (tall case)
    H = Q_old.T @ S_new
    U_small, steps_small, err_small = ns_polar_square(H, tol=tol, max_steps=120)
    Q_rel = Q_old @ U_small

    # Alignment to ground truth
    Q_true = polar_ground_truth(S_new)
    cosF = float(np.trace(Q_rel.T @ Q_true) / n)
    misalign = 1.0 - cosF
    frob_per_col = norm(Q_rel - Q_true, 'fro') / np.sqrt(n)
    orth_err = norm(np.eye(n) - Q_rel.T @ Q_rel, 'fro') / n

    print(f"m={m}, n={n}, cond_min={cond_min:g}, noise={noise_level}, tol={tol}")
    print(f"  NS steps on BIG S_new:   {steps_big:3d}  (final orth_err ~ {err_big:.2e})")
    print(f"  NS steps on SMALL H:     {steps_small:3d}  (final orth_err ~ {err_small:.2e})")
    print(f"  alignment 1-cosF ~ {misalign:.3e} ;  frob_diff/col ~ {frob_per_col:.3e} ;  orth_err(Q_rel) ~ {orth_err:.2e}")
    return steps_big, steps_small, misalign, frob_per_col, orth_err

# Suggested sweeps:
# 1) Start moderate: m=512, n=128, cond_min=1e-3, noise=0.02
# 2) Then harden:    cond_min=1e-6
# 3) Try smaller noise: 0.005, 0.01
if __name__ == "__main__":
    run_trial(m=512, n=128, cond_min=1e-3, noise_level=0.02, tol=1e-6)
    run_trial(m=512, n=128, cond_min=1e-6, noise_level=0.02, tol=1e-6)
    run_trial(m=512, n=128, cond_min=1e-6, noise_level=0.005, tol=1e-6)
```

```python
import numpy as np
from numpy.linalg import norm

rng = np.random.default_rng(0)

def rand_orthonormal_cols(m, n, rng):
    A = rng.standard_normal((m, n))
    # Only for synthetic init. If you insist on pure GEMM here too,
    # replace with a couple NS polishes on a random tall matrix.
    Q, _ = np.linalg.qr(A)
    return Q[:, :n]

def power_spectral_norm(A, iters=6):
    m, n = A.shape
    if m >= n:
        v = rng.standard_normal(n); v /= (norm(v)+1e-12)
        for _ in range(iters):
            u = A @ v; u /= (norm(u)+1e-12)
            v = A.T @ u; v /= (norm(v)+1e-12)
        return abs(float(u @ (A @ v))) + 1e-12
    else:
        u = rng.standard_normal(m); u /= (norm(u)+1e-12)
        for _ in range(iters):
            v = A.T @ u; v /= (norm(v)+1e-12)
            u = A @ v; u /= (norm(u)+1e-12)
        return abs(float(u @ (A @ v))) + 1e-12

def ns_polar_tall(A, tol=1e-6, max_steps=80):
    """Full Newton–Schulz polar for tall A (baseline)."""
    m, n = A.shape
    I = np.eye(n, dtype=A.dtype)
    s = power_spectral_norm(A, iters=6)
    X = A / s
    for k in range(1, max_steps+1):
        XtX = X.T @ X
        err = norm(I - XtX, 'fro') / n
        if err < tol: return X, k, err
        X = 0.5 * X @ (3*I - XtX)
    return X, max_steps, err

def ns_retract_tall(Y, polishes=3):
    """
    GEMM-only Stiefel retraction: a few Newton–Schulz 'polish' steps.
    Each polish counts as one NS step. No QR/SVD used.
    """
    m, n = Y.shape
    I = np.eye(n, dtype=Y.dtype)
    s = power_spectral_norm(Y, iters=4)
    Q = Y / s
    for _ in range(polishes):
        QtQ = Q.T @ Q
        Q = 0.5 * Q @ (3*I - QtQ)
    return Q

def stiefel_step_gemm_only(Q_old, S_new, c=0.5, polishes=3, backtracks=3):
    """
    One Riemannian gradient step on Stiefel using only GEMMs, then NS retraction.
    - Step dir: R = (I - Q Q^T) S_new
    - Step size: tau = c / ||S_new||_2 (with simple backtracking on trace objective)
    - Retraction: 'polishes' NS steps (<=3 recommended)
    Returns: Q_new, ns_steps_used (== polishes)
    """
    # Step size
    sS = power_spectral_norm(S_new, iters=6)
    tau = c / (sS + 1e-12)

    # Objective (to guide backtracking): tr(Q^T S_new)
    def obj(Q): return float(np.trace(Q.T @ S_new))

    Q_cur = Q_old
    base = obj(Q_cur)

    for _ in range(backtracks+1):
        # Riemannian gradient (changes column space)
        P = Q_cur.T @ S_new              # n×n
        R = S_new - Q_cur @ P            # m×n
        Y = Q_cur + tau * R              # tentative point off-manifold
        Q_try = ns_retract_tall(Y, polishes=polishes)  # <= polishes NS steps
        if obj(Q_try) >= base:           # accept if objective improved
            return Q_try, polishes
        tau *= 0.5                       # backtrack

    # If all backtracks fail, just return the last try
    return Q_try, polishes

def polar_ground_truth_via_full_NS(S_new, tol=1e-7):
    # For evaluation only (no SVD); use full NS to a tighter tol as "ground truth".
    Q_full, _, _ = ns_polar_tall(S_new, tol=tol, max_steps=200)
    return Q_full

def run_trial_stiefel(m=512, n=128, cond_min=1e-3, noise_level=0.02,
                      tol_big=1e-6, max_big_steps=120,
                      c=0.5, polishes=3, backtracks=3):
    # Build old seed with wide spectrum
    U = rand_orthonormal_cols(m,n,rng)
    V = rand_orthonormal_cols(n,n,rng)
    sig = np.logspace(np.log10(cond_min), 0.0, n)
    S_old = U @ (sig[:,None] * V.T)

    # "Last refresh": full NS polar of S_old
    Q_old, _, _ = ns_polar_tall(S_old, tol=min(1e-8, tol_big*0.1), max_steps=max_big_steps)

    # New seed with scaled Gaussian noise (relative)
    G = rng.standard_normal((m,n))
    G *= (noise_level * norm(S_old,'fro') / (norm(G,'fro') + 1e-12))
    S_new = S_old + G

    # Baseline: full NS on S_new
    Q_full, steps_big, err_big = ns_polar_tall(S_new, tol=tol_big, max_steps=max_big_steps)

    # Our method: 1 Stiefel step + NS retraction (GEMM-only). Total NS steps = polishes (<=3).
    Q_proc, ns_small = stiefel_step_gemm_only(Q_old, S_new, c=c, polishes=polishes, backtracks=backtracks)

    # Alignment vs "ground truth" (full NS result)
    ncols = n
    cosF = float(np.trace(Q_proc.T @ Q_full) / ncols)
    misalign = 1.0 - cosF
    frob_per_col = norm(Q_proc - Q_full, 'fro') / np.sqrt(ncols)
    orth_err = norm(np.eye(n) - Q_proc.T @ Q_proc, 'fro') / n

    print(f"m={m}, n={n}, cond_min={cond_min:g}, noise={noise_level}")
    print(f"  BASELINE Full-NS on S_new: steps={steps_big}, final orth_err ~ {err_big:.2e}")
    print(f"  OUR method (GEMM-only): NS polishes used = {ns_small}  (<= 3)")
    print(f"    alignment 1-cosF ~ {misalign:.3e} ; frob_diff/col ~ {frob_per_col:.3e} ; orth_err(Q_proc) ~ {orth_err:.2e}")

if __name__ == "__main__":
    # Try moderate conditioning; keep <=3 NS polishes.
    run_trial_stiefel(m=256, n=64, cond_min=1e-3, noise_level=0.02,
                      tol_big=1e-6, max_big_steps=120, c=0.6, polishes=3, backtracks=3)

    # Harder spectrum; you can lower c or add 1 more polish (still <=4).
    run_trial_stiefel(m=256, n=64, cond_min=1e-6, noise_level=0.02,
                      tol_big=1e-6, max_big_steps=120, c=0.5, polishes=3, backtracks=3)
```

```python
import numpy as np
from numpy.linalg import norm

rng = np.random.default_rng(0)

# ---------- utilities (GEMM-only) ----------
def power_spectral_norm(A, iters=6):
    m, n = A.shape
    if m >= n:
        v = rng.standard_normal(n); v /= (norm(v)+1e-12)
        for _ in range(iters):
            u = A @ v; u /= (norm(u)+1e-12)
            v = A.T @ u; v /= (norm(v)+1e-12)
        return abs(float(u @ (A @ v))) + 1e-12
    else:
        u = rng.standard_normal(m); u /= (norm(u)+1e-12)
        for _ in range(iters):
            v = A.T @ u; v /= (norm(v)+1e-12)
            u = A @ v; u /= (norm(u)+1e-12)
        return abs(float(u @ (A @ v))) + 1e-12

def ns_polar_tall(A, tol=1e-6, max_steps=120):
    """Full Newton–Schulz polar for tall A (baseline; used for evaluation)."""
    m, n = A.shape
    I = np.eye(n, dtype=A.dtype)
    s = power_spectral_norm(A, iters=6)
    X = A / s
    for k in range(1, max_steps+1):
        XtX = X.T @ X
        err = norm(I - XtX, 'fro') / n
        if err < tol: return X, k, err
        X = 0.5 * X @ (3*I - XtX)
    return X, max_steps, err

def ns_polar_square(H, steps=2):
    """Small n×n polar via fixed-count NS (no tolerance to cap steps)."""
    n = H.shape[0]
    I = np.eye(n, dtype=H.dtype)
    s = power_spectral_norm(H, iters=6)
    X = H / s
    for _ in range(steps):
        XtX = X.T @ X
        X = 0.5 * X @ (3*I - XtX)
    err = norm(np.eye(n) - (X.T @ X), 'fro') / n
    return X, steps, err

def ns_retract_tall(Y, polishes=1):
    """Stiefel retraction using 1–2 NS polishes (GEMM-only)."""
    m, n = Y.shape
    I = np.eye(n, dtype=Y.dtype)
    s = power_spectral_norm(Y, iters=4)
    Q = Y / s
    for _ in range(polishes):
        QtQ = Q.T @ Q
        Q = 0.5 * Q @ (3*I - QtQ)
    return Q

def rand_orthonormal_cols(m, n, rng):
    # Only for synthetic init (not performance-critical).
    A = rng.standard_normal((m, n))
    Q, _ = np.linalg.qr(A)  # you can replace with a few NS polishes if you insist on pure GEMM
    return Q[:, :n]

# ---------- phase-aligned Stiefel micro-steps (GEMM-only) ----------
def microstep_phase_aligned(Q, S, V_est, c=0.8, polishes=1, backtracks=4):
    """
    One Stiefel micro-step:
      R = (I - Q Q^T) S V_est
      Y = Q + tau * R, with tau ~ c / ||S||_2 and backtracking on tr(Q^T S V_est)
      Q <- NS-retract(Y) using 'polishes' NS polishes (1 is typically enough)
    """
    sS = power_spectral_norm(S, iters=6)
    tau = c / (sS + 1e-12)

    def obj(Qcur):
        # use phase-aligned objective to stabilize: tr(Q^T S V_est)
        return float(np.trace(Qcur.T @ (S @ V_est)))

    base = obj(Q)
    I = np.eye(Q.shape[1], dtype=Q.dtype)

    for _ in range(backtracks+1):
        P = Q.T @ (S @ V_est)         # n×n
        R = (S @ V_est) - Q @ P       # (I - Q Q^T) S V_est
        Y = Q + tau * R
        Q_try = ns_retract_tall(Y, polishes=polishes)  # <= 1–2 NS steps
        if obj(Q_try) >= base:
            return Q_try, polishes, tau
        tau *= 0.5
    return Q_try, polishes, tau  # last try

def relative_update_gemm_only(Q_old, S_new,
                              SMALL_NS_STEPS=2,   # 1–2 steps on small n×n polar
                              K_MICRO=3,         # number of Stiefel micro-steps
                              POLISHES_PER_STEP=1,
                              c=0.8):
    """
    Total NS steps budget = SMALL_NS_STEPS + K_MICRO * POLISHES_PER_STEP.
    Default: 2 + 3*1 = 5.
    """
    # small n×n polar for right phase (≤ SMALL_NS_STEPS)
    H2 = S_new.T @ Q_old
    V_est, small_steps_used, small_err = ns_polar_square(H2, steps=SMALL_NS_STEPS)

    Q = Q_old
    total_polishes = 0
    for _ in range(K_MICRO):
        Q, pol_used, _ = microstep_phase_aligned(Q, S_new, V_est, c=c, polishes=POLISHES_PER_STEP)
        total_polishes += pol_used

    return Q, small_steps_used, total_polishes, small_err

# ---------- experiment harness ----------
def run_trial(m=256, n=64, cond_min=1e-3, noise_level=0.02,
              SMALL_NS_STEPS=2, K_MICRO=3, POLISHES_PER_STEP=1, c=0.8):
    # old seed with wide spectrum
    U = rand_orthonormal_cols(m,n,rng)
    V = rand_orthonormal_cols(n,n,rng)
    sig = np.logspace(np.log10(cond_min), 0.0, n)
    S_old = U @ (sig[:,None] * V.T)

    # cached polar of S_old via full NS (refresh)
    Q_old, _, _ = ns_polar_tall(S_old, tol=1e-8, max_steps=200)

    # new seed
    G = rng.standard_normal((m,n))
    G *= (noise_level * norm(S_old,'fro') / (norm(G,'fro') + 1e-12))
    S_new = S_old + G

    # baseline: full NS on S_new (for target)
    Q_full, steps_big, err_big = ns_polar_tall(S_new, tol=1e-6, max_steps=200)

    # our GEMM-only relative update with ≤ 5 NS steps total
    Q_rel, small_steps, pol_steps, small_err = relative_update_gemm_only(
        Q_old, S_new,
        SMALL_NS_STEPS=SMALL_NS_STEPS,
        K_MICRO=K_MICRO,
        POLISHES_PER_STEP=POLISHES_PER_STEP,
        c=c
    )

    # alignment vs full NS (no SVD)
    ncols = n
    cosF = float(np.trace(Q_rel.T @ Q_full) / ncols)
    misalign = 1.0 - cosF
    frob_per_col = norm(Q_rel - Q_full, 'fro') / np.sqrt(ncols)
    orth_err_rel = norm(np.eye(n) - Q_rel.T @ Q_rel, 'fro') / n

    print(f"m={m}, n={n}, cond_min={cond_min:g}, noise={noise_level}")
    print(f"  BASELINE Full-NS: steps={steps_big}, final orth_err ~ {err_big:.2e}")
    print(f"  OUR method (GEMM-only): smallNS={small_steps}, micro={K_MICRO}×(polishes={POLISHES_PER_STEP}) "
          f"→ total NS = {small_steps + K_MICRO*POLISHES_PER_STEP} (≤ 5)")
    print(f"    small n×n polar orth_err ~ {small_err:.2e}")
    print(f"    alignment 1-cosF ~ {misalign:.3e} ; frob_diff/col ~ {frob_per_col:.3e} ; orth_err(Q_rel) ~ {orth_err_rel:.2e}")

if __name__ == "__main__":
    # Keep total NS steps <= 5: 2 (small) + 3*(1 polish) = 5
    run_trial(m=256, n=64, cond_min=1e-3, noise_level=0.02,
              SMALL_NS_STEPS=2, K_MICRO=3, POLISHES_PER_STEP=1, c=0.8)

    # Harder spectrum
    run_trial(m=256, n=64, cond_min=1e-6, noise_level=0.02,
              SMALL_NS_STEPS=2, K_MICRO=3, POLISHES_PER_STEP=1, c=0.7)
```
